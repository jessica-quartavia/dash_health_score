# EDA — Churn 90d (pesquisa)

Requer `exports/predictive-study/dataset_preditivo_validado.csv` e `npm run build:predictive-validated`.

Execute `run_audit.py` antes dos experimentos.

```python
from pathlib import Path
import pandas as pd
import yaml

ROOT = Path('.').resolve()
cfg = yaml.safe_load((ROOT / 'config.yaml').read_text(encoding='utf-8'))
data_dir = (ROOT / cfg['data_dir']).resolve()
df = pd.read_csv(data_dir / 'dataset_preditivo_validado.csv', sep=';', encoding='utf-8-sig')
eligible = df['elegivel_modelagem_90d'].astype(str).str.lower().eq('true')
cens = df['censura_90d'].astype(str).str.lower().eq('true')
lab = df[eligible & ~cens].copy()
lab['y'] = lab['desfecho_churn_90d'].astype(str).str.lower().eq('true').astype(int)
print('Clientes únicos (rotulados 90d):', lab['cliente_ref'].nunique())
print('Observações:', len(lab), 'Prevalência:', lab['y'].mean())
print(lab.groupby('mes_referencia')['y'].agg(['count','mean']))
```
